## 🔹 Introduction
NumPy provides powerful functions to perform statistical operations on arrays.  
These operations are widely used in data analysis, machine learning, and scientific computing.  

In this section, you will learn how to:  
- Calculate **mean, median, standard deviation, variance**.  
- Compute **correlation coefficients** between arrays.  

---

## 🔹 Creating an Array
Let’s start with a simple array.


````python
import numpy as np
````

````python
arr = np.array([10, 20, 30, 40, 50])  
print("Array:", arr)
````

---

## 🔹 Mean
The mean is the average of all elements.

````python
mean_val = np.mean(arr)  
print("Mean:", mean_val)
````

---

## 🔹 Median
The median is the middle value when the array is sorted.


````python
median_val = np.median(arr)  
print("Median:", median_val)
````

---

## 🔹 Standard Deviation
Standard deviation measures how spread out the values are.


````python
std_val = np.std(arr)  
print("Standard Deviation:", std_val)
````

---

## 🔹 Variance
Variance is the square of the standard deviation.


````python
var_val = np.var(arr)  
print("Variance:", var_val)
````

---

## 🔹 Correlation Coefficient
Correlation measures the relationship between two arrays.  
Values range from -1 (perfect negative) to +1 (perfect positive).



````python
arr1 = np.array([1, 2, 3, 4, 5])  
arr2 = np.array([2, 4, 6, 8, 10])  

corr_matrix = np.corrcoef(arr1, arr2)  
print("Correlation Coefficient Matrix:\n", corr_matrix)
````
