## 🔹 Introduction
NumPy makes mathematical operations on arrays simple and efficient.  
Instead of looping through elements manually, NumPy allows **vectorized operations** that are faster and more concise.  

In this section, you will learn how to:  
- Perform **element‑wise addition, subtraction, multiplication, and division**.  
- Calculate the **dot product** of two vectors.  
- Calculate the **cross product** of two vectors.  

---

## 🔹 Creating Arrays
Let’s start by creating two arrays of the same shape.

````python
import numpy as np
````

````python
arr1 = np.array([1, 2, 3])  
arr2 = np.array([4, 5, 6])  

print("Array 1:", arr1)  
print("Array 2:", arr2)
````

---

## 🔹 Element‑wise Operations
NumPy applies operations element by element when arrays have the same shape.

````python
print("Addition:", arr1 + arr2)  
print("Subtraction:", arr1 - arr2)  
print("Multiplication:", arr1 * arr2)  
print("Division:", arr1 / arr2)
````

---

## 🔹 Dot Product
The dot product multiplies corresponding elements and sums them.

````python
print(arr1)
print(arr2)
dot_product = np.dot(arr1, arr2)  
print("Dot Product:", dot_product)
````


---

## 🔹 Cross Product
The cross product applies only to 3‑dimensional vectors.

````python
cross_product = np.cross(arr1, arr2)  
print("Cross Product:", cross_product)
````
