````python
# Do not edit: This is your autograder code
import hashlib
import platform
import datetime

def run_tests(student_validate_password):
    # Hard-coded expected outputs
    tests = [
        ("Case 1", "MyPass123!", True),
        ("Case 2", "short1!", False),          # too short
        ("Case 3", "alllowercase123!", False), # no uppercase
        ("Case 4", "ALLUPPERCASE123!", False), # no lowercase
        ("Case 5", "NoDigitsHere!", False),    # no digit
        ("Case 6", "ValidPass9@", True),
    ]

    points_per_test = 10
    total = len(tests) * points_per_test
    score = 0
    feedback = []

    for label, pwd, expected in tests:
        try:
            student_result = student_validate_password(pwd)
            if student_result == expected:
                score += points_per_test
                feedback.append(f"✅ {label} passed")
            else:
                feedback.append(f"❌ {label} failed (got {student_result}, expected {expected})")
        except Exception as e:
            feedback.append(f"❌ {label} raised error: {e}")

    return {"score": score, "total": total, "feedback": "\n".join(feedback)}

def generate_token(assignment_tag: str):
    sysinfo = platform.node() + platform.system() + platform.release()
    now = datetime.datetime.now()
    batch_slot = (now.hour // 2)
    date_str = now.strftime("%Y-%m-%d")
    combined = sysinfo + assignment_tag + date_str + str(batch_slot)
    uid = hashlib.sha256(combined.encode()).hexdigest()[:10]
    return uid

if __name__ == "__main__":
    result = run_tests(validate_password)
    print(result['feedback'])
    print(f"Final Score: {result['score']} / {result['total']}")
    if result['score'] == result['total']:
        token = generate_token('Practical_8_Task_2')
        print('\nToken:', token)
    else:
        print('\n⚠️ Token not generated because not all tests passed.')
````
