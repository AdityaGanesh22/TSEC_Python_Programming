````python
# Do not edit: This is your autograder code
import hashlib
import platform
import datetime

def run_tests(student_validate_phone, student_validate_email):
    # Hard-coded expected outputs
    tests = [
        ("Case 1", "9876543210", "student@example.com", True, True),
        ("Case 2", "12345", "wrongemail@", False, False),
        ("Case 3", "9999999999", "hello@world.org", True, True),
        ("Case 4", "abcdefghij", "abc@xyz", False, False),
    ]

    points_per_test = 10
    total = len(tests) * 2 * points_per_test  # phone + email per case
    score = 0
    feedback = []

    for label, phone, email, expected_phone, expected_email in tests:
        try:
            student_phone = student_validate_phone(phone)
            student_email = student_validate_email(email)

            if student_phone == expected_phone:
                score += points_per_test
                feedback.append(f"✅ {label} phone passed")
            else:
                feedback.append(f"❌ {label} phone failed (got {student_phone}, expected {expected_phone})")

            if student_email == expected_email:
                score += points_per_test
                feedback.append(f"✅ {label} email passed")
            else:
                feedback.append(f"❌ {label} email failed (got {student_email}, expected {expected_email})")

        except Exception as e:
            feedback.append(f"❌ {label} raised error: {e}")

    return {"score": score, "total": total, "feedback": "\n".join(feedback)}

def generate_token(assignment_tag: str):
    sysinfo = platform.node() + platform.system() + platform.release()
    now = datetime.datetime.now()
    batch_slot = (now.hour // 2)
    date_str = now.strftime("%Y-%m-%d")
    combined = sysinfo + assignment_tag + date_str + str(batch_slot)
    uid = hashlib.sha256(combined.encode()).hexdigest()[:10]
    return uid

if __name__ == "__main__":
    # In notebook, you would call run_tests(validate_phone, validate_email)
    # Here we just show the structure
    result = run_tests(validate_phone, validate_email)
    print(result['feedback'])
    print(f"Final Score: {result['score']} / {result['total']}")
    if result['score'] == result['total']:
        token = generate_token('Practical_8_Task_1')
        print('\nToken:', token)
    else:
        print('\n⚠️ Token not generated because not all tests passed.')
````
