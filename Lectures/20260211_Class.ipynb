# File handling

The purpose is to create files, read them, add additional information to the file, etc.

1. write a file = w
2. read a file = r
3. append a file = a
4. read and write a file = r+

```python
with open("dummy_file.txt", "w") as f:
    f.write("This is the first line\n")
    f.write("This is the second line\n")
    f.write("This is the third line\n")
    f.write("This is the fourth line\n")
```

```python
with open("dummy_file_2.txt", "r") as f:
    lines = f.read().split()
    lines.sort()
    print(lines)
```

```python
with open("dummy_file_2.txt", "a") as f:
    f.write("\n \tThis is the fourth line")
```

```python
with open("dummy_file.txt", "r+") as f:
    lines = f.read()
    print(lines)
    f.write("This is the eighth line\n")
```
