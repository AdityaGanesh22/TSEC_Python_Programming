# 🧬 Bioinformatics and String Matching

In bioinformatics, DNA sequences are represented as strings made up of the four nucleotides: **A, T, G, C**. Comparing these sequences allows us to detect mutations, measure similarity, and identify motifs (short recurring patterns) that may have biological significance.

One of the simplest ways to compare two strings is by calculating the **Hamming distance**. This metric counts the number of positions at which the corresponding symbols are different. For example, if two DNA sequences differ at 3 positions, their Hamming distance is 3. This is especially useful when studying mutations or errors in DNA replication.
