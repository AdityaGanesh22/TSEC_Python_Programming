````python
# Basic arithmetic operations

print(5 + 3)             # addition(+)
print(5 - 1)             # subtraction(-)
print(5 * 3)             # multiplication(*)
print(6 / 2)             # division(/)
print(5 ** 2)            # exponential(**)
print(5 % 2)             # modulus(%) returns the remainder (1 in this case)
print(5 // 2)            # Floor division operator(//)

# Checking data types
print(type(100))                            # Int
print(type(3.14))                           # Float
print(type(True))                           # bool
print(type(1 + 2j))                         # Complex number
print(type('Python programming'))           # String
print(type([1, 2, 3]))                      # List
print(type({'name':'Python programming'}))  # Dictionary
print(type({9.81, 3.14, 2.76}))             # Set
print(type((9.81, 2.76)))             # Tuple
````

### while loops

````python
# while condition:
#     continue your logic
#     if condition is satisfied
#      break the logic
name = "concentration"
i = 0
while i < 5:
    print(f"The {name} is: ", i, "mol/L")
    i += 1
````

````python
flag = True
i = 0
while flag:
    print(f"The {name} is: ", i, "mol/L")
    i += 1
    if i >= 5:
        flag = False
````

````python
i = 0
while True:
    print(f"The {name} is: ", i, "mol/L")
    i += 1
    if i >= 5:
        break
````

````python
i = 0
while i < 3:
    j = 0
    while j < 3:
        print(i,j)
        j += 1
    i += 1
````

### functions

````python
def print_conc(n=2):
    i = 0
    name = 'Concentration'
    while True:
        print(f"The {name} is: ", i, "mol/L")
        i += 1
        if i >= n:
            break
````

## Error handling
If we expect an error in our code due to some exception, we can set up a try, except condition where it will bypass the failing state

````python
num = 10
denom = 2
try:
    print(num/denom)
except:
    print("This statement will fail: ")
````

Generate the following pattern shown in the whiteboard

so let us say that you have a function
`pattern_generator(n)`

You should generate multiple lines of the same pattern. Keep in mind that you are creating a function to even include a 100 x 100 pattern grid


Let us work on the pseudocode

```
function -> pattern_generator, input -> n
use logic to print empty spaces, then use logic to print stars, then follow it with empty spaces if need be (can be omitted as well)
end function
```

````python
def treepattern(n):
    # Here n is the number of stars in the last row of the tree
    for i in range(int(0.5*(n+1))):
        print(" " * (int(0.5*(n+1)) - i - 1) + "*" * (2 * i + 1))
treepattern(9)
````
