```python
# Do not tamper this code. Run it and then see the file that has been generated. You will have to read this file to continue the assignment.

import numpy as np

# Parameters
n_points = 200000        # number of time points (large file)
dt = 0.01                # time step
time = np.arange(0, n_points*dt, dt)

# Generate synthetic turbulent fluctuations
# Base signal: Gaussian random noise with correlation
# We'll add a sinusoidal component to mimic large-scale eddies
np.random.seed(42)       # for reproducibility
noise = np.random.normal(0, 1, n_points)
large_scale = 0.5 * np.sin(0.1 * time)

# Combine to form fluctuating u component
u_fluct = noise + large_scale

# Save to file
data = np.column_stack((time, u_fluct))
np.savetxt("u_fluctuations.txt", data, header="time u_fluct", comments="")

print("File 'u_fluctuations.txt' generated with", n_points, "rows.")
```

## 📘 Appendix 2: Autocorrelation Made Simple

Autocorrelation tells us how similar a signal is to itself when shifted in time.
Instead of thinking about all possible shifts, let’s fix a few lag values (like 1, 10, 50 steps)
and see how the signal compares to itself.

---

### Equation
If you have saved your time series as arrays `t` (time) and `u` (fluctuating velocity),
the autocorrelation at lag k is:
\begin{equation*}
    R(k) = \frac{1}{N-k}  \sum_{i=0}^{N-k-1} u_{i}u_{i+k}
\end{equation*}

where:
- N = total number of data points
- k = lag (how many steps you shift forward)

We usually normalize so that R(0) = 1.

### Interpretation
- If R(k) quickly drops to zero → the signal is mostly random noise.
- If R(k) stays large for some lags → the signal has memory.
- If R(k) oscillates → the signal has a repeating pattern.
