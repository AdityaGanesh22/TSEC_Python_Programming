# Classes

Purpose is to create a game with characters and associated powers.

```python
class Character:
    def __init__(self, name, health, damage, xp):
        self.name = name
        self.health = health
        self.damage = damage
        self.xp = xp

class Warrior(Character):
    def __init__(self, name, health, damage, xp):
        super().__init__(name, health, damage, xp)
    
    def attack(self, enemy):
        return enemy.health - self.damage

class Mage(Character):
    def __init__(self, name, health, damage, xp):
        super().__init__(name, health, damage, xp)
    
    def attack(self, enemy):
        return enemy.health - self.damage

class Medic(Character):
    def __init__(self, name, health, damage, xp):
        super().__init__(name, health, damage, xp) # here damage represents the healing power of the medic
    
    def heal(self, ally):
        return ally.health + self.damage

class Necromancer(Character):
    def __init__(self, name, health, damage, xp):
        super().__init__(name, health, damage, xp)
    
    def summon(self, enemy):
        return enemy.health - 2*self.damage

class Samurai(Warrior):
    def __init__(self, name, health, damage, xp):
        super().__init__(name, health, damage, xp)

    def attack(self, enemy):
        return enemy.health - 3*self.damage
```
