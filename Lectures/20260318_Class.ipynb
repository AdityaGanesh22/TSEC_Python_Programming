# Exploring GUI in python using tkinter

Tkinter is Python's standard library for building Graphical User Interfaces (GUIs).
In this tutorial, we will learn:
- How to create a basic window
- Adding labels, buttons, and text inputs
- Handling events (button clicks)
- Organizing widgets with layouts
- A mini project: Simple Calculator

## Creating a basic window

Task: Change the name of your window and size of your geometry

```python
import tkinter as tk

# Create a basic window
root = tk.Tk()                          # creates the main application window.
root.title("My First Tkinter Window")   # sets the window title.
root.geometry("300x200")                # sets the window size to 300 pixels wide and 200 pixels tall.

# Run the application
root.mainloop()                         # starts the event loop, keeping the window open until the user closes it.
```

## Labels and Buttons

Labels display text or images, while Buttons allow user interaction.

```python
root = tk.Tk()
root.title("Labels and Buttons")

# Create a label widget
label = tk.Label(root, text="Hello, Tkinter!", font=("Arial", 14))
label.pack(pady=10)  # pack() places the widget, pady adds vertical spacing

# Create a button widget
button = tk.Button(root, text="Click Me")
button.pack(pady=10)

root.mainloop()
```

## Handling Events

We can make buttons interactive by attaching functions to them.

```python
root = tk.Tk()
root.title("Event Handling")

# Define a function to run when button is clicked
def on_click():
    # Update label text when button is pressed
    label.config(text="Button Clicked!")

# Create a label
label = tk.Label(root, text="Waiting...", font=("Arial", 14))
label.pack(pady=10)

# Create a button with an event handler
button = tk.Button(root, text="Click Me", command=on_click)
button.pack(pady=10)

root.mainloop()
```

## Text Entry

Entry widgets allow users to type text input.

```python
root = tk.Tk()
root.title("Text Entry Example")

def show_text():
    # Get text from entry widget
    user_text = entry.get()
    # Update label with user input
    label.config(text=f"You typed: {user_text}")

# Create an entry widget
entry = tk.Entry(root, width=20)
entry.pack(pady=10)

# Button to trigger show_text function
button = tk.Button(root, text="Submit", command=show_text)
button.pack(pady=10)

# Label to display user input
label = tk.Label(root, text="")
label.pack(pady=10)

root.mainloop()
```

## Layout Management

Tkinter provides three geometry managers:
- **pack()**: simple stacking
- **grid()**: table-like layout
- **place()**: absolute positioning

```python
root = tk.Tk()
root.title("Layout Example")

# Using grid layout (row, column)
tk.Label(root, text="Row 0, Col 0").grid(row=0, column=0)
tk.Label(root, text="Row 0, Col 1").grid(row=0, column=1)
tk.Label(root, text="Row 1, Col 0").grid(row=1, column=0)
tk.Label(root, text="Row 1, Col 1").grid(row=1, column=1)

root.mainloop()
```

## Mini Project: Simple Calculator

Let's combine what we've learned to build a basic calculator.

```python
root = tk.Tk()
root.title("Simple Calculator")

def calculate():
    try:
        # Evaluate the expression entered by the user
        result = eval(entry.get())
        label.config(text=f"Result: {result}")
    except Exception as e:
        # Handle invalid input gracefully
        label.config(text="Error!")

# Entry widget for user input
entry = tk.Entry(root, width=20)
entry.pack(pady=10)

# Button to trigger calculation
button = tk.Button(root, text="Calculate", command=calculate)
button.pack(pady=10)

# Label to display result
label = tk.Label(root, text="Result: ")
label.pack(pady=10)

root.mainloop()
```

# Regular Expressions (Regex)
## Introduction to Regex

Regular Expressions (regex) are patterns used to match text.  
They are extremely useful for:
- Searching text
- Validating input (like emails, phone numbers)
- Extracting specific parts of strings
- Replacing text

We use Python's built-in `re` module for regex operations.

Key functions:
- `re.match()` → checks for a match at the beginning of a string
- `re.search()` → searches for a match anywhere in the string
- `re.findall()` → returns all matches as a list
- `re.sub()` → replaces matched text

```python
import re

# Example: search for "cat" in a sentence
text = "The cat sat on the mat."
match = re.search("cat", text)

if match:
    print("Found:", match.group())  # group() returns the matched text
```

# Basic Patterns

- `.` → matches any single character
- `\d` → matches any digit (0-9)
- `\w` → matches any word character (letters, digits, underscore)
- `\s` → matches whitespace (space, tab, newline)

Quantifiers:
- `+` → one or more
- `*` → zero or more
- `?` → zero or one
- `{n}` → exactly n times

```python
text = "My number is 12345."

# Find digits
digits = re.findall(r"\d+", text)  # r"" means raw string (important for regex)
print("Digits found:", digits)

# Find words
words = re.findall(r"\w+", text)
print("Words found:", words)
```

## Anchors and Boundaries

- `^` → matches start of string
- `$` → matches end of string
- `\b` → word boundary

```python
text = "hello world"

# Match only if string starts with "hello"
print(re.match(r"^hello", text))

# Match only if string ends with "world"
print(re.search(r"world$", text))
```

```python
text = "cat scatter category"

# Without \b, it matches inside other words too
matches_no_boundary = re.findall(r"cat", text)
print(matches_no_boundary)  # Output: ['cat', 'cat', 'cat']

# Match 'cat' only as a whole word
matches = re.findall(r"\bcat\b", text)
print(matches)  # Output: ['cat']
```

## Character Classes and Alternation

- `[abc]` → matches 'a', 'b', or 'c'
- `[0-9]` → matches any digit
- `[^abc]` → matches anything except 'a', 'b', or 'c'
- `|` → OR operator

```python
text = "I like cats and dogs."

# Match either 'cat' or 'dog'
animals = re.findall(r"cat|dog", text)
print("Animals found:", animals)

# Match vowels
vowels = re.findall(r"[aeiou]", text)
print("Vowels found:", vowels)
```

## Groups and Capturing

- `( )` → captures part of the match
- `group(n)` → returns nth captured group

```python
text = "My email is student@example.com"

# Capture username and domain separately
match = re.search(r"(\w+)@(\w+\.\w+)", text)
if match:
    print("Username:", match.group(1))
    print("Domain:", match.group(2))
```

## Substitution

- `re.sub(pattern, replacement, text)` replaces matched text.

```python
text = "I have 3 cats and 2 dogs."

# Replace digits with '#'
new_text = re.sub(r"\d", "#", text)
print(new_text)
```
