## Arithmatic Progression

```python
# 6,9,12,15,....
def AP(a,d,n):
    sum = 0
    list = []
    for i in range (n):
        sum = a + d*i
        list.append(sum)
    return list
a = AP(6,3,5)
print(a)
```

```python
#RPS

import random

def rps():
    hand = ["rock","paper","scissors"]

    print("Welcome to the game of ROCK, PAPER, SCISSORS")
    print("0.ROCK , 1.PAPER , 2.SCISSORS")
    cp = 0
    p1 = 0
    pwin = 0
    cwin = 0
    tie = 0
    n=3
    i=0
    for i in range(n):
        cp = random.randint(0,2)
        p1=int(input("Enter your choice: "))
        if (p1==cp):
            print("TIE!")
            tie+=1
            print(hand[p1]," ",hand[cp])
        elif(p1==0 and cp==1):
            print("Computer WINS!")
            print(hand[p1]," ",hand[cp])
            cwin+=1
        elif(p1==1 and cp==2):
            print("Computer WINS!")
            print(hand[p1]," ",hand[cp])
            cwin+=1
        elif(p1==2 and cp==0):
            print("Computer WINS!")
            print(hand[p1]," ",hand[cp])
            cwin+=1
        elif(p1==1 and cp==0):
            print("Player WINS!")
            print(hand[p1]," ",hand[cp])
            pwin+=1
        elif(p1==2 and cp==1):
            print("Player WINS!")
            print(hand[p1]," ",hand[cp])
            pwin+=1
        elif(p1==0 and cp==2):
            print("Player WINS!")
            print(hand[p1]," ",hand[cp])
            pwin+=1
    print("Player wins",pwin)
    print("Computer wins",cwin)
    print("ties", tie)

rps()
```
