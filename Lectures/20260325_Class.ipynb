## 🔹 Inner, Outer, and Dot Products

### Dot Product (Inner Product)
The dot product of two vectors \($\mathbf{u}$, $\mathbf{v}$ in $\mathbb{R}^n$\) is defined as:  

\begin{align*}
\mathbf{u} \cdot \mathbf{v} &= \sum_{i=1}^n u_i v_i \\
\mathbf{u}^T \mathbf{v} = \begin{bmatrix}u_1 & u_2 & u_3\end{bmatrix} \cdot \begin{bmatrix}v_1 \\ v_2 \\ v_3\end{bmatrix} &= u_1 v_1 + u_2 v_2 + u_3 v_3
\end{align*}

#### Example


```python
u = np.array([2, 3, 4])  
v = np.array([1, 0, -1])  
inner = np.dot(u, v)
print("Dot Product:", inner)
```

### Outer Product
The outer product ($\mathbf{u} \otimes \mathbf{v}$) of two 3×1 vectors produces a 3×3 matrix:  

\begin{align*}
\mathbf{u} \mathbf{v}^T = \begin{bmatrix}u_1 \\ u_2 \\ u_3\end{bmatrix} \cdot \begin{bmatrix}v_1 & v_2 & v_3\end{bmatrix} =
\begin{bmatrix}
u_1 v_1 & u_1 v_2 & u_1 v_3 \\
u_2 v_1 & u_2 v_2 & u_2 v_3 \\
u_3 v_1 & u_3 v_2 & u_3 v_3
\end{bmatrix}
\end{align*}


```python
u = np.array([2, 3, 4])  
v = np.array([1, 0, -1]) 
outer = np.outer(u, v)  
print("Outer Product:\n", outer) 
```

### Double Dot Product (Tensor Contraction)
For two second‑order tensors $\mathcal{A}, \mathcal{B} \in \mathbb{R}^{n \times n}$:  

\begin{align*}
\mathcal{A} : \mathcal{B} &= \sum_{i=1}^n \sum_{j=1}^n A_{ij} B_{ij} \\
&= trace(\mathcal{AB}^T)
\end{align*}

Fluid mechanics often requires:  
- Stress tensor $\sigma_{ij}$.  
- Strain rate tensor $\varepsilon_{ij}$.  
- Operations like $\sigma : \varepsilon$ (double dot product).  

### Example: Stress–Strain Contraction

\begin{equation*}
\mathrm{Energy} = \sigma : \varepsilon = \sum_{i=1}^n \sum_{j=1}^n \sigma_{ij} \varepsilon_{ij}
\end{equation*}


```python
# Example:  
A = np.array([[1,2],[3,4]])  
B = np.array([[5,6],[7,8]])  
double_dot = np.tensordot(A, B, axes=2)  
print("Double Dot Product:", double_dot)  
```

## 🔹 Curl of a Vector Field
The curl of a vector field $\mathbf{F} = (u,v,w)$ is defined as:  

\begin{equation*}
\nabla \times \mathbf{F} =
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
\frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\
u & v & w
\end{vmatrix}
\end{equation*}

For a 2D velocity field $(u(x,y), v(x,y))$, the curl reduces to:  

\begin{equation*}
\text{curl} = \frac{\partial v}{\partial x} - \frac{\partial u}{\partial y}
\end{equation*}

### Example: Circular Flow

```python
import numpy as np  
import matplotlib.pyplot as plt  

x, y = np.meshgrid(np.linspace(-2,2,20), np.linspace(-2,2,20))  
u = -y   # velocity in x-direction  
v = x    # velocity in y-direction  

dv_dx = np.gradient(v, axis=1)  
du_dy = np.gradient(u, axis=0)  
curl = dv_dx - du_dy  

plt.figure(figsize=(6,6))  
plt.quiver(x, y, u, v)  
plt.title("Velocity Field (Circular Flow)")  
plt.show()
```

# 📘 Practical Session – Symbolic Mathematics with SymPy  

---

## 🔹 Introduction
SymPy is a Python library for **symbolic mathematics**. Unlike NumPy (numerical), SymPy works with exact algebraic expressions.  
It can:  
- Simplify expressions.  
- Perform differentiation and integration.  
- Solve algebraic and differential equations.  
- Work with matrices and linear algebra symbolically.  

---

---

## 🔹 Basic Symbolic Variables
We first define symbols to represent variables.


```python
from sympy import symbols, pprint
x, y = symbols('x y')  

expr = x**2 + 2*x*y + y**2  
print("Expression:", expr)
display("Expression:", expr)
```

---

## 🔹 Simplification
SymPy can simplify algebraic expressions.


```python
from sympy import expand, factor  
from IPython.display import display

expanded = expand((x + y)**2)  
factored = factor(x**2 + 2*x*y + y**2)  

display("Expanded:",expanded)
display("Factored:",factored)
```

---

## 🔹 Differentiation
SymPy can compute derivatives symbolically.


```python
from sympy import diff  

f = x**3 + 2*x**2 + x  
df = diff(f, x)  
display("Derivative:", df)
```

---

## 🔹 Integration
SymPy can compute indefinite and definite integrals.


```python
from sympy import integrate  

f = x**2  
indef = integrate(f, x)  
definite = integrate(f, (x, 0, 2))  

display("Indefinite Integral:", indef)  
display("Definite Integral (0 to 2):", definite)
```

---

## 🔹 Solving Equations
SymPy can solve algebraic equations exactly.


```python
from sympy import solve  

eq = x**2 - 4  
solutions = solve(eq, x)  
display("Solutions:", solutions)
```

---

## 🔹 Matrices and Linear Algebra
SymPy can handle matrices symbolically.


```python
from sympy import Matrix  

A = Matrix([[1, 2], [3, 4]])  
display("Matrix A:", A)  
display("Determinant:", A.det())  
display("Inverse:", A.inv())
```

---

## 🔹 Differential Equations
SymPy can solve ordinary differential equations (ODEs).

```python
from sympy import Function, dsolve, Eq  
from sympy.abc import t  

f = Function('f')  
ode = Eq(f(t).diff(t, t) - f(t), 0)  
solution = dsolve(ode)  
display("ODE Solution:", solution)
```

# 📘 Practical Session – Accelerating Python with Numba  

---

## 🔹 Introduction
Numba is a **just‑in‑time (JIT) compiler** for Python that translates numerical functions into fast machine code.  
It is especially useful for:  
- Loops over large arrays.  
- Numerical computations.  
- Scientific simulations.  

With Numba, code that takes **seconds** in pure Python can run **almost instantly**.

---

## 🔹 Installing Numba
If Numba is not installed, run:  

```python
!pip install numba
```

---

## 🔹 Example: Pure Python vs Numba

We’ll compute a heavy numerical task: summing squares in a loop.  

### Pure Python Version


```python
import time

def slow_sum_squares(n):
    total = 0
    for i in range(n):
        total += i*i
    return total

start = time.time()
result = slow_sum_squares(10**7)
end = time.time()

print("Result:", result)
print("Time (Python):", end - start, "seconds")
```

---

### Numba Optimized Version

```python
from numba import njit
```

```python
@njit
def fast_sum_squares(n):
    total = 0
    for i in range(n):
        total += i*i
    return total

start = time.time()
result = fast_sum_squares(10**7)
end = time.time()

print("Result:", result)
print("Time (Numba):", end - start, "seconds")
```
