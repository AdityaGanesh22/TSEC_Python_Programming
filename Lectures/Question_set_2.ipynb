## 🧬 GC Skew Analysis in Circular DNA

### Why is GC Skew Important?
In bacteria such as *E. coli*, the genome is **circular DNA**. DNA replication begins at a specific location called the **origin of replication**. Identifying this origin is crucial because:
- It helps us understand how DNA replication is regulated.
- It reveals asymmetries in nucleotide composition between the leading and lagging strands.
- It is a fundamental bioinformatics technique used in genome analysis.

One way to detect the origin is by computing the **GC skew** across the genome:
- Each `G` contributes **+1**.
- Each `C` contributes **–1**.
- `A` and `T` contribute **0**.
- The skew is computed cumulatively across the sequence.
- The **minimum point of the skew curve** often corresponds to the origin of replication.
