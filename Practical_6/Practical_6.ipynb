# Vehicle Rental System – Class Structure

- **Vehicle (Base Class)**
  - `__init__(vehicle_id, model, base_rate)`
    - requires: vehicle ID (string)
    - requires: model name (string)
    - requires: base rental rate (float)
    - initializes: availability status (True by default)
  - `get_id()`
    - returns: vehicle ID
  - `get_model()`
    - returns: model name
  - `is_available()`
    - returns: True/False
  - `mark_unavailable()`
    - action: sets availability to False
  - `mark_available()`
    - action: sets availability to True
  - `calculate_price(days)`
    - requires: rental period in days (int)
    - returns: base_rate * days

- **Car (Subclass of Vehicle)**
  - `calculate_price(days)`
    - returns: base_rate * days * 1.2 (20% surcharge)

- **Bike (Subclass of Vehicle)**
  - `calculate_price(days)`
    - returns: base_rate * days * 0.8 (20% discount)

- **Truck (Subclass of Vehicle)**
  - `calculate_price(days)`
    - returns: base_rate * days * 1.5 (50% surcharge)

- **RentalTransaction (Class)**
  - `__init__(customer_name, vehicle, days)`
    - requires: customer name (string)
    - requires: Vehicle object
    - requires: rental period in days (int)
    - calculates: total cost using vehicle.calculate_price()
  - `process()`
    - action: confirms rental if vehicle available, marks vehicle unavailable, prints summary

- **RentalAgency (Class)**
  - `__init__(name)`
    - requires: agency name (string)
    - initializes: empty vehicle list
  - `add_vehicle(vehicle)`
    - requires: Vehicle object
    - action: adds vehicle to agency inventory
  - `list_available_vehicles()`
    - action: prints all available vehicles
  - `rent_vehicle(customer_name, vehicle_id, days)`
    - requires: customer name (string), vehicle ID (string), rental period (int)
    - action: creates RentalTransaction and processes booking
